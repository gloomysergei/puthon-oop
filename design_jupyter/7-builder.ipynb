#### src/solution.py

Уже реализован `класс Pizza`, который содержит следующие поля:

- `size` — размер пиццы, строка
- `dough` — тип теста, строка   `dough [dəu] - тесто`
- `sauce` — соус, строка         `sauce [sɔ:s] - соус`
- `meat_topping` — мясная начинка, строка
- `vegetable_topping` — овощная начинка, строка
- `cheese_topping` — сорт сыра, строка

Реализуйте в нём билдер `PizzaBuilder`. Билдер должен задавать ингридиенты и возвращать новый экземпляр `Pizza` в метод get_pizza.

Реализуйте функцию `create_classic_pizza()`, которая сконструирует и вернет пиццу "Маргарита":
- большого размера `(big)`
- на тонком тесте `(thin)`
- сыром Моцарелла `(mozzarella)`. 
- Соус у пиццы томатный `(tomato)`,
- в качестве овощной начинки используем базилик `(basil)`. 
При создании пиццы в функции `create_classic_pizza()` используйте этот билдер.

```python
pizza = create_classic_pizza()

pizza.get_size()  # big
pizza.get_dough()  # thin
pizza.get_sauce()  # tomato
pizza.get_vegetable_topping()  # basil
pizza.get_cheese_topping()  # mozzarella
```
