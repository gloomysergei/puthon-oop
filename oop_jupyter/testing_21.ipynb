#### Классические алгоритмы обхода BST
---

##### Структура дерева
```text

        9
       / \
      4   17
     / \    \
    8   6    22
       / \   /
      3   7 20
```
---


##### 1. In-order (слева → узел → справа)

**Порядок:** `8, 4, 3, 6, 7, 9, 17, 20, 22`

**Когда используют:** для BST это даёт значения по возрастанию. В твоём случае дерево не является корректным BST (из‑за узла 8 слева от 4), поэтому сортированного порядка не будет, но сам алгоритм тот же.

**Рекурсивно:**

```python
def inorder(node):
    if node is None:
        return []
    return inorder(node.left) + [node.key] + inorder(node.right)
```
**Итеративно (через стек):**

```python
def inorder_iter(root):
    res, stack = [], []
    cur = root
    while cur or stack:
        while cur:
            stack.append(cur)
            cur = cur.left
        cur = stack.pop()
        res.append(cur.key)
        cur = cur.right
    return res
```
##### 2. Pre-order (узел → слева → справа)

Порядок: 9, 4, 8, 6, 3, 7, 17, 22, 20

**Когда используют:** чтобы «сериализовать» дерево, сделать копию, вывести структуру сверху вниз.

**Рекурсивно:**

```python
def preorder(node):
    if node is None:
        return []
    return [node.key] + preorder(node.left) + preorder(node.right)
```
Итеративно (через стек, корень первым):
python

def preorder_iter(root):
    if not root:
        return []
    res, stack = [], [root]
    while stack:
        node = stack.pop()
        res.append(node.key)
        # сначала кладём правый, чтобы левый был сверху
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    return res

3. Post-order (слева → справа → узел)

Порядок: 8, 3, 7, 6, 4, 20, 22, 17, 9

Когда используют: удаление дерева, вычисление выражений в дереве, агрегация снизу вверх.

Рекурсивно:
python

def postorder(node):
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.key]

Итеративно (два стека или один стек с реверсом):
python

def postorder_iter(root):
    if not root:
        return []
    stack, res = [root], []
    while stack:
        node = stack.pop()
        res.append(node.key)
        if node.left:
            stack.append(node.left)
        if node.right:
            stack.append(node.right)
    # порядок получается обратный (корень в конце), поэтому разворачиваем
    return res[::-1]

4. Level-order (по уровням, слева направо)

Порядок: 9, 4, 17, 8, 6, 22, 3, 7, 20

Когда используют: поиск кратчайшего пути в дереве, BFS, вывод «как нарисовано».

Реализация через очередь:
python

from collections import deque

def level_order(root):
    if not root:
        return []
    q = deque([root])
    res = []
    while q:
        node = q.popleft()
        res.append(node.key)
        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)
    return res

Как применить к твоему тесту
python

tree = Node(
    9,
    Node(
        4,
        Node(8),
        Node(
            6,
            Node(3),
            Node(7),
        ),
    ),
    Node(
        17,
        right=Node(
            22,
            Node(20),
        ),
    ),
)

print(inorder(tree))        # [8, 4, 3, 6, 7, 9, 17, 20, 22]
print(preorder(tree))       # [9, 4, 8, 6, 3, 7, 17, 22, 20]
print(postorder(tree))      # [8, 3, 7, 6, 4, 20, 22, 17, 9]
print(level_order(tree))    # [9, 4, 17, 8, 6, 22, 3, 7, 20]
